# import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load the dataset

```python
df = pd.read_csv('myexcel - myexcel.csv.csv')
```

# 1. Preprocessing

```python
np.random.seed(42)
df['Height'] = np.random.randint(150, 181, size=len(df))

df['Salary'] = df['Salary'].fillna(df['Salary'].median())

print("Preprocessing Complete: Height corrected and missing Salaries imputed.\n")
```

# 2. Analysis & Visualization

```python
# Task 1: Team Distribution
team_counts = df['Team'].value_counts()
team_pct = (team_counts / len(df)) * 100
team_summary = pd.DataFrame({'Count': team_counts, 'Percentage': team_pct})

plt.figure(figsize=(12, 6))
sns.barplot(x=team_counts.index, y=team_counts.values, palette='viridis')
plt.xticks(rotation=90)
plt.title('Task 1: Employee Distribution by Team')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()
```

```python
# Task 2: Position Segregation
pos_counts = df['Position'].value_counts()

plt.figure(figsize=(8, 5))
pos_counts.plot(kind='pie', autopct='%1.1f%%', colors=sns.color_palette('pastel'))
plt.title('Task 2: Employee Segregation by Position')
plt.ylabel('')
plt.show()
```

```python
# Task 3: Predominant Age Group
# Creating bins for ages
bins = [20, 25, 30, 35, 40, 45]
labels = ['20-25', '25-30', '30-35', '35-40', '40-45']
df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)
age_dist = df['Age Group'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
sns.countplot(x='Age Group', data=df, palette='flare')
plt.title('Task 3: Predominant Age Group')
plt.show()
```

```python
# Task 4: Highest Salary Expenditure (Team & Position)
team_exp = df.groupby('Team')['Salary'].sum().sort_values(ascending=False)
pos_exp = df.groupby('Position')['Salary'].sum().sort_values(ascending=False)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
team_exp.head(10).plot(kind='bar', ax=ax1, color='teal')
ax1.set_title('Top 10 Teams by Salary Spending')
pos_exp.plot(kind='bar', ax=ax2, color='coral')
ax2.set_title('Salary Spending by Position')
plt.tight_layout()
plt.show()
```

```python
# Task 5: Age vs Salary Correlation
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Salary', data=df, hue='Position', alpha=0.7)
sns.regplot(x='Age', y='Salary', data=df, scatter=False, color='black')
plt.title('Task 5: Correlation between Age and Salary')
plt.show()

correlation = df['Age'].corr(df['Salary'])
print(f"Correlation Coefficient: {correlation:.2f}")
```

# 3. Final Summary 

```python
print("\n--- Data Insights ---")
print(f"1. Most employees are in the {age_dist.idxmax()} age group.")
print(f"2. The {team_exp.idxmax()} has the highest salary expenditure.")
print(f"3. The most common position is {pos_counts.idxmax()}.")
print(f"4. Correlation of {correlation:.2f} indicates age has a small positive impact on salary.")
```
